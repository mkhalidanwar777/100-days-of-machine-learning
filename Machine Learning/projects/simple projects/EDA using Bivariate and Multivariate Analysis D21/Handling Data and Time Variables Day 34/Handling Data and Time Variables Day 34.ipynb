```python
import numpy as np
import pandas as pd
```

```python
date = pd.read_csv('orders.csv')
time = pd.read_csv('messages.csv')
```

# Date

```python
date.head()
```

```python
date.info()
```

## Now change the Datatype

```python
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

# 1 Extract Year

```python
date['date_year']=date['date'].dt.year
```

```python
date.head()
```

# 2 Extract Month

```python
date['date_month']=date['date'].dt.month
date.head()
```

## 2.1 Extract Month in english

```python
date['date_month_name']=date['date'].dt.month_name()
date.head()
```

# 3 Extract DAy

```python
date['date_day']=date['date'].dt.day
date.head()
```

## 3.1 Extract Day name in english

```python
date['date_day_name']=date['date'].dt.day_name()
date.head()
```

```python
date.drop(columns=['product_id','city_id','orders'],inplace=True)
```

```python
date.head()
```

## 3.2 day of Week

```python
date['date_day_of_week']=date['date'].dt.dayofweek
date.head()
```

# 4 WeekEnd 

```python
date['date_WeekEnd']=np.where(date['date_day_name'].isin(['Saturday','Sunday']),1,0)
date.head()
```

# 5 Exract Week of the Year

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.head()
```

# 6 Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter
date.head()
```
